```python


def return_search(name='',cb='',verbose=False):
    # Replace this name with any central banker of interest
    decisionmaker_name = name + ',' + cb#"Abdellatif Jouahri, Bank Al-Maghrib"
    #"Vittorio Corbo, Central Bank of Chile"
    #"Peter Mooslechner, Austrian National Bank"
    
    # Define a structured prompt that requests a JSON output with sources
    query = f"""
    You are a meticulous researcher with access to web search.
    
    I want you to gather the educational background of the following person:
    
    **{decisionmaker_name}**
    
    Please return the results as a JSON object with the following structure:
    
    {{
      "Date_of_brith": "",
      "Bachelor_institution": "",
      "Bachelor_year": "",
      "PhD_institution": "",
      "PhD_year": "",
      "PhD_thesis_director": "",
      "highest_level_of_study": "",
      "Sources": {{
        "Date_of_birth": "",
        "Bachelor_institution": "",
        "Bachelor_year": "",
        "PhD_institution": "",
        "PhD_year": "",
        "PhD_thesis_director": "",
        "highest_level_of_study": ""
      }}
    }}
    
    Instructions:
    - Fill in all known educational institution.
    - For each field, provide a direct source URL in the `"Sources"` object.
    - Use credible and verifiable sources such as official biographies, CVs, academic databases, or reputable media.
    - If a field is not publicly available or cannot be verified, leave it as an empty string (`""`).
    - Do not add any explanatory text before or after the JSON output.
    - If you cannot find any information in a given entry return None for that entry
    """
    
    # Create the response using the web search tool
    response = client_oai.responses.create(
        model="gpt-4o",
        tools=[{"type": "web_search_preview",
               "search_context_size": "high",
               }],
        #response_format="json",
        input=query
    )
    
    # Print the output
    if verbose == True:
      print("🔎 Question:", query)
      print("\n📄 JSON Answer:")
      print(response.output_text)
    return response.output[-1].content[0].text
```
